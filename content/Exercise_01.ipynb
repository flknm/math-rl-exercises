## 1. Discounted returns

Consider a finite reward sequence $r_0, r_1, \ldots, r_{T-1}$. Its discounted return is

$$
G_0 = \sum_{t=0}^{T-1} \gamma^t r_t,
$$

where the discount factor satisfies $\gamma \in [0,1]$.

**Example.** We use four rewards and $\gamma = 0.9$.

```python
rewards = [1, 0, 2, 1]
gamma = 0.9

G0 = sum(gamma**t * reward for t, reward in enumerate(rewards))
print(f"Discounted return: {G0:.4f}")
```
